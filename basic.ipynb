```python
 #write a program in narcissistic number
def is_narcissistic(num):
    # Convert number to string to easily iterate over digits
    digits = str(num)
    power = len(digits)  # number of digits
    total = sum(int(digit) ** power for digit in digits)
    return total == num

# Example usage
numbers = [153, 9474, 123, 407, 1634]
for n in numbers:
    if is_narcissistic(n):
        print(f"{n} is a Narcissistic number")
    else:
        print(f"{n} is NOT a Narcissistic number")
```
