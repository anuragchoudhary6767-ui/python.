```python
#WAP which will print sum of all the keys whos value are the factor of 12.
dt={1:2,2:3,3:4,6:12,17:18,18:12}
ans=0
for k,v in dt.items():
    if 12%v==0:
        ans+=k

print(ans)
```

```python
# Given dictionary
dt = {1:2, 2:3, 3:4, 6:12, 17:18, 18:12}

# Initialize sum
total = 0

# Loop through dictionary
for key, value in dt.items():
    if 12 % value == 0:   # check if value is factor of 12
        total += key

print("Sum of keys whose values are factors of 12 =", total)
```
